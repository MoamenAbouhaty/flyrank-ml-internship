## 4 The unit of analysis, as a real dataframe

**One row = one content page.**

Each `content_id` represents a content page belonging to a client.

The prediction/scoring decision is therefore made at the page level: how strongly should this page be prioritized for content refresh review?

```python
import pandas as pd

df = pd.read_csv("/content/content_refresh_anonymized.csv")

print("Shape:", df.shape)

display(
    df[
        [
            "content_id",
            "client_id",
            "impressions_90d",
            "impressions_last_30d",
            "impressions_prev_30d",
            "clicks_90d",
            "ctr",
            "avg_position",
            "trend_direction",
            "trend_pct"
        ]
    ].head(10)
)
```

## 5) Why ML beats a fixed rule here

A fixed rule could say:

> "Prioritize every page with at least 500 impressions and a declining trend."

This is a useful baseline, but it treats all qualifying pages similarly.

An ML scoring approach could combine multiple signals, including impressions, clicks, CTR, average position, recent performance, content age, freshness, search volume, and engagement.

This can produce a ranked list instead of a simple yes/no decision.

The fixed rule should remain as a baseline. ML is useful if it can produce better ranking quality and identify valuable opportunities that a simple threshold misses.

The final output supports a real content action: deciding which pages should be reviewed first for a content refresh.

```python
# Create a simple refresh-opportunity proxy

df["is_declining"] = df["trend_direction"].eq("down")

df["refresh_opportunity_proxy"] = (
    df["is_declining"]
    & (df["impressions_90d"] >= 500)
).astype(int)

display(
    df[
        [
            "content_id",
            "impressions_90d",
            "trend_direction",
            "trend_pct",
            "refresh_opportunity_proxy"
        ]
    ].head(10)
)

print(
    "Proxy opportunities:",
    df["refresh_opportunity_proxy"].sum()
)
```
