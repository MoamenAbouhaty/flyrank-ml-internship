## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains **30,000 pages** and **44 columns**.

In the dataset, **16,262 pages** are labeled as declining based on the `trend_direction` field.

Also, **16,726 pages** have at least **500 impressions over the last 90 days**.

These numbers give an initial view of the dataset and show that both declining pages and pages with meaningful search visibility are common enough to support further analysis.


```python
import os
import pandas as pd

csv_path = None

for root, dirs, files in os.walk("/content"):
    if "content_refresh_anonymized.csv" in files:
        csv_path = os.path.join(root, "content_refresh_anonymized.csv")
        break

print("CSV:", csv_path)

if csv_path is None:
    print("Error: 'content_refresh_anonymized.csv' not found. Please upload the file or ensure it's in the correct directory.")
else:
    df = pd.read_csv(csv_path)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    print("\nDeclining pages:")
    print((df["trend_direction"] == "down").sum())

    print("\nPages with 500+ impressions:")
    print((df["impressions_90d"] >= 500).sum())
```
